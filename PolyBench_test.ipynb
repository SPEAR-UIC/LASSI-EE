```python
import subprocess
```

### Compile

```python
output_file = "my_bench"

cmd = [
            "gcc",
            "-O3",
            "-I", "source_codes/PolyBenchC-4.2.1/utilities",
            "-I", "source_codes/PolyBenchC-4.2.1/linear-algebra/kernels/atax",
            "source_codes/PolyBenchC-4.2.1/utilities/polybench.c",
            "source_codes/PolyBenchC-4.2.1/linear-algebra/kernels/atax/atax.c",
            "-DPOLYBENCH_TIME",
            "-DEXTRALARGE_DATASET",
            "-o", str(output_file),
        ]

print(" ".join(cmd))

result = subprocess.run(cmd, capture_output=True, text=True)
```

### Run

```python
output = subprocess.run(
    ["./source_codes/PolyBenchC-4.2.1/utilities/time_benchmark.sh", "./my_bench"],
    check=True,
)
```
