```python
import aiohttp
import asyncio
import json
import nest_asyncio
import sys

# Allow nested async loops inside Jupyter
nest_asyncio.apply()

async def force_flush():
    """Continuously flush stdout to ensure real-time updates in Jupyter"""
    while True:
        await asyncio.sleep(0.1)  # Small sleep to yield control
        sys.stdout.flush()

async def fetch_stream():
    url = "https://localhost:5000/argoapi/api/v1/resource/streamchat/"
    headers = {
        "Content-Type": "application/json",
        "Accept": "text/plain",
        "Host": "apps-dev.inside.anl.gov"  # Ensure correct API hostname
    }
    
    data = {
        "user": "mdearingtest",
        "model": "gpt4o",
        "system": "You are a large language model with the name Argo.",
        "prompt": ["Tell me something interesting about quantum mechanics."],
        "stop": [],
        "temperature": 0.2,
        "top_p": 0.9
    }

    # Start a background flush task
    flush_task = asyncio.create_task(force_flush())

    async with aiohttp.ClientSession() as session:
        async with session.post(url, json=data, headers=headers, ssl=False) as response:
            print(f"Status: {response.status}")
            print("Streaming Response:")

            async for chunk in response.content.iter_any():
                text = chunk.decode(errors="replace")
                
                sys.stdout.write(text)  # Write directly to stdout
                sys.stdout.flush()  # Ensure immediate printing
                
                await asyncio.sleep(0)  # Yield to event loop for display updates

    # Cancel the background flush task once done
    flush_task.cancel()

# Run the async function inside Jupyter Notebook
await fetch_stream()
```

```python
import subprocess

def fetch_with_curl():
    url = "https://localhost:5000/argoapi/api/v1/resource/streamchat/"
    headers = [
        "-H", "Content-Type: application/json",
        "-H", "Host: apps-dev.inside.anl.gov"
    ]
    data = '{"user": "mdearingtest", "model": "gpt4o", "system": "You are a large language model with the name Argo.", "prompt": ["Tell me something interesting about quantum mechanics."], "stop": [], "temperature": 0.2, "top_p": 0.9}'

    # Run curl as a subprocess and capture real-time output
    process = subprocess.Popen(
        ["curl", "-k", "-X", "POST", url, *headers, "-d", data],
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE,
        text=True
    )

    output = ""
    
    # Read output line by line to stream it live
    for line in process.stdout:
        print(line, end="", flush=True)  # Live print each token
        output += line  # Store full response

    return output  # Store full response in case you need to use it

# Run the function and capture the output
output = fetch_with_curl()
print("\nFinal Output:", output)
```
